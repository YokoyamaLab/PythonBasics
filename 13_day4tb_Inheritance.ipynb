## 🌀差分プログラミングとは

差分プログラミングとは、既存のプログラムやクラスを継承し、必要な部分だけを独自に実装することで、プログラムの作成を効率化する手法のことです。継承により、継承元の機能をそのまま利用し、継承先で独自の機能を追加することで、コードの再利用性を高め、開発効率を向上させる事ができます。

## 🌀クラスの継承

まずは以下のコードを見てください。

````python
class MyPortrait(SelfPortrait):

    def drawEyes(self):
        major_r = 10
        miner_r = 5
        eye_right_center = (150,220)
        eye_left_center = (250,220)
        eye_right = (
            (eye_right_center[0]-major_r , eye_right_center[1]-miner_r),
            (eye_right_center[0]+major_r , eye_right_center[1]+miner_r)
        )
        eye_left = (
            (eye_left_center[0]-major_r,eye_left_center[1]-miner_r),
            (eye_left_center[0]+major_r,eye_left_center[1]+miner_r)
        )
        #白目を楕円で描画
        self.canvas.ellipse(eye_right,fill=(255,255,255))
        self.canvas.ellipse(eye_left,fill=(255,255,255))
        #黒目を円で描画
        self.canvas.circle(eye_right_center,miner_r,fill=self.color_eye)
        self.canvas.circle(eye_left_center,miner_r,fill=self.color_eye)
        self.canvas.circle(eye_right_center,miner_r-4,fill=(0,0,0))
        self.canvas.circle(eye_left_center,miner_r-4,fill=(0,0,0))

my_portrait = MyPortrait("Umeko Tsuda",color_eye=(25, 68, 142))
image = my_portrait.draw(debug=False)
image
````

**MyPortrait**というクラスに```drawEyes()```のみが定義されています。しかしクラス定義の部分が少々異なります。

```python
class MyPortrait(SelfPortrait):
```

これは**SelfPortraitクラス**を継承した**MyPortraitクラス**という意味です。この時**SelfPortraitクラス**を**親クラス**、**MyPortraitクラス**を**子クラス**ど呼びます。

**子クラス**のインスタンスを作成すると、そのインスタンスは**親クラス**のすべてのメソッドやクラス変数、インスタンス変数を引き継いだ上で、**子クラス**に定義されたメソッドを持ちます。**子クラス**のメソッドで**親クラス**と同名のものがあれば、**子クラス**が優先されます。これをオーバライドと呼びます。

上記のコードを実行すると、**目**以外は**親クラス**で描画されたものと同じように描画されています。これが**差分プログラミング**です。

では**親クラス**のメソッド全体をオーバライドするのではなく、**親クラス**の描画を活かした上で、さらに加筆したい場合はどうすれば良いのでしょうか。

以下の例は、親クラスで目を描画した後で、そこにキラキラを追加しています。

````python
class MyPortrait(SelfPortrait):

    def drawEyes(self):
        super().drawEyes()
        xy = (159,225)
        r= 2
        self.canvas.circle(xy,r,fill=(255,255,255))
        xy = (239,225)
        self.canvas.circle(xy,r,fill=(255,255,255))
        xy = (163,218)
        r= 3
        self.canvas.circle(xy,r,fill=(255,255,255))
        xy = (243,218)
        self.canvas.circle(xy,r,fill=(255,255,255))



my_portrait = MyPortrait("Umeko Tsuda")
image = my_portrait.draw(debug=False)
image
````

コードを見てみましょう。

```python
    def drawEyes(self):
        super().drawEyes()
        xy = (159,225)
        # この後キラキラを追記するコード
```

見慣れないのは```super()```ですね。これは**親クラス**を返す関数です。つまり親クラスの```drawEyes()```で目を描かせた後、独自コードでキラキラを追加しています。こうなるとキラキラの色を変えたくなりますね？色はコンストラクタで指定していますので、そこに```color_kirakira```を追加しましょう。

コンストラクタは親クラスのをそのままコピーしてきて、引数の後ろに```color_kirakira```を追加します。デフォルトの色も決めておきましょう。

また、```drawEyes()```の中で色の指定を```(255,255,255)```から```self.color_kirakira```に変更しています。これでキラキラが指定した色で描かれるはずです。

では、インスタンス作成時にコンストラクタに```color_kirakira```に桜色を指定して実行しましょう。どうでしょうか？

````python
class MyPortrait(SelfPortrait):
    def __init__(
        self,
        name,
        color_skin=(255, 219, 172),
        color_hair=(26, 27, 33),
        color_lip=(220, 20, 60),
        color_eye=(12, 12, 45),
        color_nose=(0, 0, 0),
        color_background=(236, 239, 241),
        color_name=(0, 0, 0),
        color_kirakira=(187, 228, 250)
    ):
        super().__init__(name,color_skin,color_hair,color_lip,color_eye,color_nose,color_background,color_name)
        self.color_kirakira = color_kirakira

    def drawEyes(self):
        super().drawEyes()
        xy = (159,225)
        r= 2
        self.canvas.circle(xy,r,fill=self.color_kirakira)
        xy = (239,225)
        self.canvas.circle(xy,r,fill=self.color_kirakira)
        xy = (163,218)
        r= 3
        self.canvas.circle(xy,r,fill=self.color_kirakira)
        xy = (243,218)
        self.canvas.circle(xy,r,fill=self.color_kirakira)

my_portrait = MyPortrait("Umeko Tsuda",color_kirakira=(250, 187, 238))
image = my_portrait.draw(debug=False)
image
````
