### 💮長方形

【[リファレンス](https://pillow.readthedocs.io/en/stable/reference/ImageDraw.html#PIL.ImageDraw.ImageDraw.rectangle)】

```python
# 画像の大きさと背景色を決める(色は赤,緑,青の強さを0～255で表す)
image = Image.new("RGBA", (400, 400), (0, 0, 255))

# 図形を描くキャンバスを準備する
canvas = ImageDraw.Draw(image)

# キャンバスに長方形を描画する
canvas.rectangle(
    [(50, 100), (300, 350)],  # 左上のxy座標、右下のXY座標
    fill=(255, 0, 0),         # 塗りの色
    outline=(0, 255, 0),      # 線の色
    width=10                  # 線の太さ
)

# colabページ上に表示する
image
```

```python
# 画像の大きさと背景色を決める(色は赤,緑,青の強さを0～255で表す)
image = Image.new("RGBA", (400, 400), (0, 0, 255))

# 図形を描くキャンバスを準備する
canvas = ImageDraw.Draw(image)

# キャンバスに線分を描画する
canvas.line(
    [(50, 100), (300, 350),(50, 350),(280, 20)],  # 左上のxy座標、右下のXY座標
    fill=(0,255,0),
    width=5
)

# colabページ上に表示する
image
```

### 💮ポリゴン

[リファレンス](https://pillow.readthedocs.io/en/stable/reference/ImageDraw.html#PIL.ImageDraw.ImageDraw.polygon)

```python
# 画像の大きさと背景色を決める(色は赤,緑,青の強さを0～255で表す)
image = Image.new("RGBA", (400, 400), (0, 0, 255))

# 図形を描くキャンバスを準備する
canvas = ImageDraw.Draw(image)

# キャンバスに線分を描画する
canvas.polygon(
    [(50, 100), (300, 350),(50, 350),(280, 20)],  # 左上のxy座標、右下のXY座標
    fill=(0,255,0),
    outline=(0,40,0)
)

# colabページ上に表示する
image
```

### 💮多角形

【[リファレンス](https://pillow.readthedocs.io/en/stable/reference/ImageDraw.html#PIL.ImageDraw.ImageDraw.regular_polygon)】

```python
# 画像の大きさと背景色を決める(色は赤,緑,青の強さを0～255で表す)
image = Image.new("RGBA", (400, 400), (0, 0, 255))

# 図形を描くキャンバスを準備する
canvas = ImageDraw.Draw(image)

steps = 10
# for文で設定を変えながら沢山描いてみる
for i in range(steps):
  # キャンバスに多角形を描画する
  canvas.regular_polygon(
      (200,200,200-(i*15)),
      n_sides = steps + 3 - i,
      rotation = i * 360 / steps,
      fill=(0,25 * i,0)
  )

# colabページ上に表示する
image
```

### 💮楕円／円

【[リファレンス](https://pillow.readthedocs.io/en/stable/reference/ImageDraw.html#PIL.ImageDraw.ImageDraw.ellipse)】

```python
# 画像の大きさと背景色を決める(色は赤,緑,青の強さを0～255で表す)
image = Image.new("RGBA", (400, 400), (0, 0, 255))

# 図形を描くキャンバスを準備する
canvas = ImageDraw.Draw(image)

# キャンバスに長方形を描画する
canvas.rectangle(
    [(50, 100), (300, 350)],  # 左上のxy座標、右下のXY座標
    outline=(255, 255, 255),  # 線の色
    width=1                   # 線の太さ
)

# キャンバスに楕円を描画する（上記長方形に内接する円が描ける）
canvas.ellipse(
    [(50, 100), (300, 350)],  # 左上のxy座標、右下のXY座標
    outline=(0, 255, 0),      # 線の色
    fill=(255,0,0),           # 塗りの色
    width=3                   # 線の太さ
)


# colabページ上に表示する
image
```

### 💮円弧

円弧は端点の処理方法で三種類あります。

【[リファレンス: arc](https://pillow.readthedocs.io/en/stable/reference/ImageDraw.html#PIL.ImageDraw.ImageDraw.arc)】

【[リファレンス: chord](https://pillow.readthedocs.io/en/stable/reference/ImageDraw.html#PIL.ImageDraw.ImageDraw.chord)】

【[リファレンス: pieslice](https://pillow.readthedocs.io/en/stable/reference/ImageDraw.html#PIL.ImageDraw.ImageDraw.pieslice)】

```python
# 画像の大きさと背景色を決める(色は赤,緑,青の強さを0～255で表す)
image = Image.new("RGBA", (400, 400), (0, 0, 255))

# 図形を描くキャンバスを準備する
canvas = ImageDraw.Draw(image)
coordinates = [(50, 100), (300, 350)]

# キャンバスに長方形を描画する
canvas.rectangle(
    coordinates,  # 左上のxy座標、右下のXY座標
    outline=(255, 255, 255),  # 線の色
    width=1                   # 線の太さ
)

# 赤がarc
canvas.arc(
    coordinates,  # 左上のxy座標、右下のXY座標
    start=5,
    end=115,
    fill=(255,0,0),
    width=3
)

# 緑がchord
canvas.chord(
    coordinates,
    start=125,
    end=235,
    fill=(0,255,0),
    outline=(0,120,0),
    width=3
)

# 黄色がpieslice
canvas.pieslice(
    coordinates,  # 左上のxy座標、右下のXY座標
    start=245,
    end=355,
    fill=(255,255,0),
    outline=(120,120,0),
    width=3
)
# colabページ上に表示する
image
```

### 💮文字を描く

【[リファレンス](https://pillow.readthedocs.io/en/stable/reference/ImageDraw.html#PIL.ImageDraw.ImageDraw.text)】

ASCII文字(数字、アルファベット、基本的な記号など)以外の文字、例えばひらがな・カタカナ・漢字を描画したい場合は、フォントを追加する必要があります。

```python
!wget -P /usr/share/fonts/ https://github.com/YokoyamaLab/PythonBasics/raw/refs/heads/2025/font/NotoSansCJKjp-DemiLight.ttf
cjk_font_path = "/usr/share/fonts/NotoSansCJKjp-DemiLight.ttf"
```

```python
# 画像の大きさと背景色を決める(色は赤,緑,青の強さを0～255で表す)
image = Image.new("RGBA", (400, 400), (244, 81, 30))

# 図形を描くキャンバスを準備する
canvas = ImageDraw.Draw(image)

# キャンバスに文字列を描画する
cjk_font = ImageFont.truetype(cjk_font_path, 32)
canvas.text(
    (10,20),
    "ハローワールド",
    fill=(248, 187, 208),
    font=cjk_font,
    spacing=4,
    align='left')

# colabページ上に表示する
image
```
