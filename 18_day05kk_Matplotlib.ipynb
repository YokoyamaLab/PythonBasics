### 💮[1本目] $y=x^2$のグラフ

まずは単純な折れ線グラフを書いてみましょう。

[plot関数のリファレンス](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html)

````python
# 0～10までの連番を生成
x = np.arange(0, 11)
# それを二乗したものをyにセット
y = x ** 2

## ここからがグラフ描画部

# 散布図はscatter()関数
plt.plot(x, y)

# グラフ全体と各軸にラベル
plt.xlabel('x')
plt.ylabel('y')
plt.title('y = x^2')
plt.grid(True)

# グラフ描画！
plt.show()
````

### 💮[2本目] ランダムデータで散布図のグラフ

散布図はデータ可視化の基本です。それではランダムなデータを生成して散布図にしてみましょう。

````python
num_points = 200
random_x = np.random.rand(num_points) * 10  # random x values between 0 and 10
random_y = np.random.rand(num_points) * 100 # random y values between 0 and 100

# 散布図はscatter()関数
plt.scatter(random_x, random_y)

# グラフ全体と各軸にラベル
plt.xlabel('ランダムなx')
plt.ylabel('ランダムなy')
plt.title('ランダムなデータの散布図')

# わかりやすいようにグリッド線も描画する
plt.grid(True)

# グラフ描画！
plt.show()
````

### 💮[4本目] 複数のデータ描画と凡例

次に複数のデータを一つのグラフに入れ込む例を考えましょう。その場合混乱をさせるために凡例の表示も必要になります。

````python
# データ①
x = np.arange(0, 11)
y = x ** 2

# データ②
x2 = np.arange(0, 11)
y2 = 2 * x2 + 5

# 二種類のデータをセット（ここのlabelが凡例に表れる）
plt.plot(x, y, label='y = x^2')
plt.plot(x2, y2, label='y = 2x + 5')

# グラフ全体と各軸にラベル
plt.xlabel('x')
plt.ylabel('y')
plt.title('Multiple Datasets')

# 凡例を付ける
plt.legend()

# グリッドは描画する
plt.grid(True)

# グラフ描画！
plt.show()
````

GitHub上のプログラミング言語のシェア(Top8)

| 順位 | 言語       | シェア率       |
|------|------------|----------------|
| 1    | JavaScript | 0.245          |
| 2    | Python     | 0.165          |
| 3    | Java       | 0.085          |
| 4    | TypeScript | 0.080          |
| 5    | C++        | 0.065          |
| 6    | PHP        | 0.055          |
| 7    | C#         | 0.050          |

````python
languages_share = [
    {"language": "JavaScript", "share": 0.245},
    {"language": "Python", "share": 0.165},
    {"language": "Java", "share": 0.085},
    {"language": "TypeScript", "share": 0.080},
    {"language": "C++", "share": 0.065},
    {"language": "PHP", "share": 0.055},
    {"language": "C#", "share": 0.050},
    {"language": "C", "share": 0.045}
]

# ラベルとデータに分離
labels = [d['language'] for d in languages_share]
share = [d['share'] for d in languages_share]

# その他の生成(1からシェアの合計を引く)
labels.append('その他')
share.append(1 - sum(share))

# 描画領域は正方形
plt.figure(figsize=(8, 8))

# 円グラフを指定(%表示(autopct)、真上から時計回り(startangle,counterclock)を指定)
plt.pie(share, labels=labels, autopct='%1.1f%%', startangle=90, counterclock=False)
plt.title('プログラミング言語のシェア')

# グラフ描画！
plt.show()

````

### 💮[11本目] グラフのアニメーション

````python
# アニメーションの表示に必要な機能のimport
from matplotlib import animation
from IPython.display import HTML

# アニメーションの長さ(滑らかさ)
num_frames = 100
x_anim = np.linspace(0, 10, num_frames)
y_anim = x_anim ** 2

# グラフの定義
fig, ax = plt.subplots()
ax.set_xlim(0, 10)
ax.set_ylim(0, 100)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('y = x^2 Animation')
ax.grid(True)

# 初期(空っぽ)のグラフ作成
line, = ax.plot([], [], 'r-')

# アニメーション終了後の初期化
def init():
  line.set_data([], [])
  return line,

# iフレーム目の描画
def animate(i):
  line.set_data(x_anim[:i], y_anim[:i])
  return line,

# アニメーション作成
anim = animation.FuncAnimation(fig, animate, init_func=init,
                               frames=num_frames, interval=50, blit=True)

# 動画にして表示
HTML(anim.to_html5_video())
````

### 💮[12本目] インタラクティブなグラフの生成

グラフはインタラクティブにする事も可能です。ipywidgetsはcolabo上でボタンやスライダーといったGUI部品を使うための機能で、これを用いてリアルタイムに描画するグラフに影響を与える事が可能です。

````python
# Colab上でGUI部品を使えるようにする機能をimport
from ipywidgets import interact, FloatSlider

# サンプルデータ（このデータをスライダーで操作します）
x_interactive = np.linspace(0, 10, 100)
y_interactive = np.sin(x_interactive)

# グラフを更新する関数
def update_plot(amplitude, frequency, phase):
    plt.figure(figsize=(8, 6))
    plt.plot(x_interactive, amplitude * np.sin(frequency * x_interactive + phase))
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title('Interactive Sine Wave')
    plt.grid(True)
    plt.ylim(-2, 2) # Y軸の範囲を固定
    plt.show()

# インタラクティブウィジェットを作成
# FloatSliderはスライダーウィジェットです。
# (初期値, 最小値, 最大値, ステップ)
interact(update_plot,
         amplitude=FloatSlider(value=1.0, min=0.1, max=2.0, step=0.1, description='振幅:'),
         frequency=FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description='周波数:'),
         phase=FloatSlider(value=0.0, min=0.0, max=2*np.pi, step=0.1, description='位相:'));

````

### 💮[10本目] トレンド(移動平均)の生成

グラフはデータから描画するだけでなく、データを分析した結果も重ねて表示する事が出来ます。

````python
# 10時間分の株価データ(ランダムなダミーデータ)を10分間隔で生成
np.random.seed(0)
dates = pd.date_range('2023-01-01 09:00', periods=600, freq='1T')
stock_prices = np.random.normal(loc=100, scale=10, size=len(dates))

# データフレームを作成
df = pd.DataFrame({'Date': dates, 'Stock Price': stock_prices})
df.set_index('Date', inplace=True)

# 移動平均を計算
df['Moving Average'] = df['Stock Price'].rolling(window=10).mean()

# 10分間隔ごとの箱ひげ図と移動平均のグラフを描画
plt.figure(figsize=(12, 6))

# 箱ひげ図
plt.boxplot([df['Stock Price'][i:i+10] for i in range(0, len(df), 10)])


# 移動平均
plt.plot(range(1, 61), [df['Moving Average'][i*10] for i in range(60)], color='red', label='Moving Average')

# 題名・軸ラベル等
plt.title('Box Plot of Stock Prices (10-minute intervals) with Moving Average')
plt.xlabel('10-minute intervals')
plt.ylabel('Price')
plt.xticks(ticks=range(1, 61), labels=[f'{i*10}-{(i+1)*10} min' for i in range(60)], rotation=90)
plt.legend()
plt.tight_layout()

plt.show()
````

なんとなく相関がありそうなグラフが出てきました。ではこのデータに対して最小二乗法で回帰直線を引いてみましょう。詳細は次回以降で説明しますが、scikit-learnという機械学習のライブラリに線形回帰のツールがありますので、それを利用します。

````python
# 機械学習のライブラリscikit-learnをimport (これについての詳細は次回)
from sklearn.linear_model import LinearRegression

# 日の出時刻のデータから経度と日の出時刻を抽出し、NumPy配列に変換
X_sunrise = np.array(longitudes).reshape(-1, 1) # 経度を特徴量として使用 (scikit-learnは2D配列を期待するためreshape)
y_sunrise = np.array(sunrises)              # 日の出時刻をターゲット変数として使用

# 線形回帰モデルをインスタンス化
model_sunrise = LinearRegression()

# モデルをデータにフィット（学習=最小二乗法）させる
model_sunrise.fit(X_sunrise, y_sunrise)

# フィットしたモデルから勾配 (coef_) と切片 (intercept_) を取得
slope_sunrise = model_sunrise.coef_[0]    # 勾配
intercept_sunrise = model_sunrise.intercept_ # 切片

# 回帰直線を描くためのyの値を計算
# y_pred = 勾配 * X + 切片
# 回帰直線をプロットするために、X_sunriseの範囲で直線上の点を計算します。
X_fit_sunrise = np.linspace(X_sunrise.min(), X_sunrise.max(), 100).reshape(-1, 1)
y_fit_sunrise = model_sunrise.predict(X_fit_sunrise)

# 散布図と回帰直線をプロット
plt.figure(figsize=(10, 6)) # グラフのサイズを設定
plt.scatter(X_sunrise, y_sunrise, label='データポイント (都市)') # 元のデータポイントを散布図で表示
plt.plot(X_fit_sunrise, y_fit_sunrise, color='red', label=f'回帰直線 (y = {slope_sunrise:.2f}x + {intercept_sunrise:.2f})') # 計算した回帰直線をプロット

# ラベルとタイトルを追加
plt.xlabel('経度')
plt.ylabel('日の出時刻')
plt.title('経度と日の出時刻の関係における回帰直線')

# 凡例を表示
plt.legend()

# グリッドを表示
plt.grid(True)

# グラフを表示
plt.show()

print(f"回帰直線の式: 日の出時刻 ≈ {slope_sunrise:.2f} * 経度 + {intercept_sunrise:.2f}")
````

このように、綺麗な回帰直線を引くことができました。

### 💮[16本目] バイオリン

matplotlibとseabornでは同じ事が出来ますが、コードの簡潔さが違います。『映える』チャートのバイオリン図を例に、タイタニック号での生死を年齢別で図示します。

````python
#matplotlibでヴァイオリンプロットを描く
x1 = df[df['survived'] == 0]['age'].values
x2 = df[df['survived'] == 1]['age'].values
x1 = x1[~np.isnan(x1)].tolist() # 欠損値を削除
x2 = x2[~np.isnan(x2)].tolist() # 欠損地を削除
fig = plt.figure()
ax = fig.add_subplot()
ax.violinplot([x1, x2])
ax.set_xticks([1,2])
ax.set_xticklabels(['0', '1'])
ax.set_xlabel('survived')
ax.set_ylabel('age')
plt.show()

#seabornでヴァイオリンプロットを描く
sns.violinplot(x='survived', y='age', data=df)
plt.show()
````

次にタイタニック号に乗っている乗客の客室クラスと年齢を可視化します。ファーストクラスの方が年齢層が高い事が分かります。

````python
sns.violinplot(data=df, x='pclass', y='age')
````

生死で色分けしてみましょう。

````python
fig,ax=plt.subplots()
sns.violinplot(data=df, x='pclass', y='age',hue="survived",split=True, ax=ax)
ax.legend(loc='upper left',bbox_to_anchor=(1.05,1))
handler, label = ax.get_legend_handles_labels()
ax.legend(handler, ["dead","alive"],loc='upper left',bbox_to_anchor=(1.05,1))
````
