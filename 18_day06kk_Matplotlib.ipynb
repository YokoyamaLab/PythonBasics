### 💮[1本目] $y=x^2$のグラフ

まずは単純な折れ線グラフを書いてみましょう。

[plot関数のリファレンス](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html)

```python
# 0～10までの連番を生成
x = np.arange(0, 11)
# それを二乗したものをyにセット
y = x ** 2

## ここからがグラフ描画部

# 散布図はscatter()関数
plt.plot(x, y)

# グラフ全体と各軸にラベル
plt.xlabel('x')
plt.ylabel('y')
plt.title('y = x^2')
plt.grid(True)

# グラフ描画！
plt.show()
```

### 💮[2本目] ランダムデータで散布図のグラフ

散布図はデータ可視化の基本です。それではランダムなデータを生成して散布図にしてみましょう。

```python
num_points = 200
random_x = np.random.rand(num_points) * 10  # random x values between 0 and 10
random_y = np.random.rand(num_points) * 100 # random y values between 0 and 100

# 散布図はscatter()関数
plt.scatter(random_x, random_y)

# グラフ全体と各軸にラベル
plt.xlabel('ランダムなx')
plt.ylabel('ランダムなy')
plt.title('ランダムなデータの散布図')

# わかりやすいようにグリッド線も描画する
plt.grid(True)

# グラフ描画！
plt.show()
```

### 💮[3本目] シンプルな棒グラフ

カテゴリとカテゴリ毎の数値というデータから、それぞれのカテゴリを棒グラフで可視化します。

```python
# 描画するデータを準備
categories = ['東京:', '横浜', '大阪', '名古屋']
values = [14043239, 3771063, 2791907,  2331264]

# グラフ描画領域の縦横比を事前に指定することも出来る(Option)
plt.figure(figsize=(4, 4))

# 棒グラフはbar()関数
plt.bar(categories, values, color='skyblue')

# グラフ全体と各軸にラベル
plt.xlabel('都市')
plt.ylabel('人口')
plt.title('都市と人口の棒グラフ')

# グリッドを軸(axis)や線種(linestyle)、透明度(alpha)を指定して描画も可能
plt.grid(axis='y', linestyle='--', alpha=0.7)

# グラフ描画！
plt.show()
```

### 💮[4本目] 複数のデータ描画と凡例

次に複数のデータを一つのグラフに入れ込む例を考えましょう。その場合混乱をさせるために凡例の表示も必要になります。

```python
# データ①
x = np.arange(0, 11)
y = x ** 2

# データ②
x2 = np.arange(0, 11)
y2 = 2 * x2 + 5

# 二種類のデータをセット（ここのlabelが凡例に表れる）
plt.plot(x, y, label='y = x^2')
plt.plot(x2, y2, label='y = 2x + 5')

# グラフ全体と各軸にラベル
plt.xlabel('x')
plt.ylabel('y')
plt.title('Multiple Datasets')

# 凡例を付ける
plt.legend()

# グリッドは描画する
plt.grid(True)

# グラフ描画！
plt.show()
```

### 💮[5本目] ガウス分布のデータの可視化(ヒストグラム)



```python
# ガウス分布に従うデータの生成
mu, sigma = 0, 0.1
gaussian_data = np.random.normal(mu, sigma, 10000)

# ヒストグラムはhist()関数、この例では色とエッジ色を指定しています(option)
plt.hist(gaussian_data, bins=50,color='lightcoral', edgecolor='black')

# グラフ全体と各軸にラベル
plt.xlabel('値')
plt.ylabel('頻度')
plt.title('ガウス分布に従うデータのヒストグラム')

# グリッドはy軸のみ描画する
plt.grid(axis='y', linestyle='--', alpha=0.7)

# グラフ描画！
plt.show()
```

### 💮[7本目] 円グラフ

以下はGitHubが2023年に公開したプログラミング言語のシェアを基に、一部、演習のために、数値を調整して作成したサンプルデータです。これを円グラフにして、プログラミング言語のシェアを直感的に理解できるようにしましょう。

GitHub上のプログラミング言語のシェア(Top8)

| 順位 | 言語       | シェア率       |
|------|------------|----------------|
| 1    | JavaScript | 0.245          |
| 2    | Python     | 0.165          |
| 3    | Java       | 0.085          |
| 4    | TypeScript | 0.080          |
| 5    | C++        | 0.065          |
| 6    | PHP        | 0.055          |
| 7    | C#         | 0.050          |
| 8    | C          | 0.045          |


```python
languages_share = [
    {"language": "JavaScript", "share": 0.245},
    {"language": "Python", "share": 0.165},
    {"language": "Java", "share": 0.085},
    {"language": "TypeScript", "share": 0.080},
    {"language": "C++", "share": 0.065},
    {"language": "PHP", "share": 0.055},
    {"language": "C#", "share": 0.050},
    {"language": "C", "share": 0.045}
]

# ラベルとデータに分離
labels = [d['language'] for d in languages_share]
share = [d['share'] for d in languages_share]

# その他の生成(1からシェアの合計を引く)
labels.append('その他')
share.append(1 - sum(share))

# 描画領域は正方形
plt.figure(figsize=(8, 8))

# 円グラフを指定(%表示(autopct)、真上から時計回り(startangle,counterclock)を指定)
plt.pie(share, labels=labels, autopct='%1.1f%%', startangle=90, counterclock=False)
plt.title('プログラミング言語のシェア')

# グラフ描画！
plt.show()

```

### 💮[8本目] 複数のグラフを一つのエリアに(サブプロット)

```python
# 2×2の4つのグラフからなるグラフを描画する
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# ここで使うデータは上記の演習で定義したデータを使います
# そのコードブロックを実行していないと、変数未定義のエラーがでます

# Plot 1: 折れ線グラフ (y = x^2)
axes[0, 0].plot(x, y)
axes[0, 0].set_xlabel('x')
axes[0, 0].set_ylabel('y')
axes[0, 0].set_title('y = x^2')
axes[0, 0].grid(True)

# Plot 2: ランダムな散布図
axes[0, 1].scatter(random_x, random_y)
axes[0, 1].set_xlabel('ランダムなx')
axes[0, 1].set_ylabel('ランダムなy')
axes[0, 1].set_title('ランダムなデータの散布図')
axes[0, 1].grid(True)

# Plot 3: 棒グラフ
axes[1, 0].bar(categories, values, color='skyblue')
axes[1, 0].set_xlabel('カテゴリ')
axes[1, 0].set_ylabel('値')
axes[1, 0].set_title('人口の棒グラフ')
axes[1, 0].grid(axis='y', linestyle='--', alpha=0.7)

# Plot 4: ヒストグラム
axes[1, 1].hist(gaussian_data, bins=30, color='lightcoral', edgecolor='black')
axes[1, 1].set_xlabel('値')
axes[1, 1].set_ylabel('頻度')
axes[1, 1].set_title('ガウス分布データのヒストグラム')
axes[1, 1].grid(axis='y', linestyle='--', alpha=0.7)

# Adjust layout to prevent overlap
plt.tight_layout()

# Show the plots
plt.show()
```

### 💮[10本目] 三次元散布図の描画

三次元グラフを描画する場合は「mpl_toolkits」から「Axes3D」をimportします。

```python
from mpl_toolkits.mplot3d import Axes3D

# 3次元プロット用のダミーデータを作成
x_3d = np.random.rand(100) * 10
y_3d = np.random.rand(100) * 10
z_3d = np.sin(x_3d) + np.cos(y_3d) + np.random.rand(100) * 2 # zはxとyの関数にノイズを追加

# 3Dプロット用のfigureとaxesを作成
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# 散布図として3次元データをプロット
ax.scatter(x_3d, y_3d, z_3d, c=z_3d, cmap='viridis', marker='o')

# ラベルとタイトルを追加
ax.set_xlabel('X軸')
ax.set_ylabel('Y軸')
ax.set_zlabel('Z軸')
ax.set_title('3次元データの散布図')

# 以下のコメントを外して視点(仰角=elev, 方位角=azim)を変更できます
# ax.view_init(elev=10, azim=80)

# グラフ描画！
plt.show()
```

### 💮[11本目] グラフのアニメーション

```python
# アニメーションに必要な機能をimportします
from matplotlib import animation
from IPython.display import HTML

# アニメーションの設定
num_frames = 100
x_anim = np.linspace(0, 10, num_frames)
y_anim = x_anim ** 2

# 冒頭でやったy = x^2をアニメーションします。
fig, ax = plt.subplots()
ax.set_xlim(0, 10)
ax.set_ylim(0, 100)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('y = x^2 Animation')
ax.grid(True)

# 初期のプロット(データは空)
line, = ax.plot([], [], 'r-')

# 初期化するため一度データを空にする関数
def init():
  line.set_data([], [])
  return line,

# 引数でフレーム数に応じて配列の一部を返す
def animate(i):
  line.set_data(x_anim[:i], y_anim[:i])
  return line,

# アニメーションの生成
anim = animation.FuncAnimation(fig, animate, init_func=init,
                               frames=num_frames, interval=50, blit=True)

# アニメーションの表示
HTML(anim.to_html5_video())
```
