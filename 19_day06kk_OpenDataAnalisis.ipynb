```python
import pandas as pd

df = pd.read_csv('mntemsadext00_rct.csv', encoding="shift-jis") #データによってはshift-jisの代わりにutf-8と入れる必要がある場合もあります

# 沖縄県と青森県のデータのみを利用
df_okinawa_aommori = df[df['都道府県'].isin(['沖縄県','青森県'])].reset_index()
df_okinawa_aommori
```
